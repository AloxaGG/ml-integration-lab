# Новый эксперимент: проверка по профилю

Функция возвращает все нарушения сразу. Сервер преобразует этот результат в согласованный HTTP 422.


```python
PROFILE = {
    'sepal_length': {'min': 4.3, 'max': 7.9, 'mean': 5.84},
    'sepal_width': {'min': 2.0, 'max': 4.4, 'mean': 3.05},
    'petal_length': {'min': 1.0, 'max': 6.9, 'mean': 3.76},
    'petal_width': {'min': 0.1, 'max': 2.5, 'mean': 1.20},
}
```

```python
def validate_feature_ranges(features, profile=PROFILE):
    errors = []
    for name, limits in profile.items():
        value = features[name]
        if not limits['min'] <= value <= limits['max']:
            errors.append({'feature': name, 'value': value, 'min': limits['min'], 'max': limits['max']})
    return errors

valid = {'sepal_length': 5.1, 'sepal_width': 3.5, 'petal_length': 1.4, 'petal_width': .2}
invalid = {**valid, 'petal_width': 4.0}
assert validate_feature_ranges(valid) == []
assert validate_feature_ranges(invalid)[0]['feature'] == 'petal_width'
```
